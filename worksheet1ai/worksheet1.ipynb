## Task 1: Image Loading & Basic Operations

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os

# Load test images (cameraman.png and lena_gray.gif)
# These are standard test images used in image processing
def load_image(path):
    """Load image and convert to numpy array."""
    img = Image.open(path).convert('L')  # Convert to grayscale
    return np.array(img)

# If images exist, load them; otherwise create synthetic test images
if os.path.exists('cameraman.png'):
    img_cameraman = load_image('cameraman.png')
else:
    # Synthetic test image
    np.random.seed(42)
    img_cameraman = np.random.randint(0, 256, (256, 256), dtype=np.uint8)
    print('Note: Using synthetic image (cameraman.png not found)')

print(f'Image shape: {img_cameraman.shape}')
print(f'Pixel range: [{img_cameraman.min()}, {img_cameraman.max()}]')
print(f'Mean intensity: {img_cameraman.mean():.2f}')

plt.figure(figsize=(6, 5))
plt.imshow(img_cameraman, cmap='gray')
plt.title('Cameraman Test Image')
plt.colorbar(label='Pixel Intensity')
plt.axis('off')
plt.show()
```

## Task 2: Image Histogram & Equalisation

```python
from skimage import exposure

# Histogram equalisation
img_eq = exposure.equalize_hist(img_cameraman)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].imshow(img_cameraman, cmap='gray')
axes[0, 0].set_title('Original Image')
axes[0, 0].axis('off')

axes[0, 1].imshow(img_eq, cmap='gray')
axes[0, 1].set_title('Histogram Equalised')
axes[0, 1].axis('off')

axes[1, 0].hist(img_cameraman.ravel(), bins=256, color='#0EA5E9', alpha=0.7)
axes[1, 0].set_title('Original Histogram')
axes[1, 0].set_xlabel('Pixel Value')

axes[1, 1].hist(img_eq.ravel(), bins=256, color='#10B981', alpha=0.7)
axes[1, 1].set_title('Equalised Histogram')
axes[1, 1].set_xlabel('Pixel Value')

plt.tight_layout()
plt.show()
```

## Task 3: Image Filtering

```python
from scipy import ndimage
from skimage.filters import gaussian, sobel

# Apply different filters
img_blur = gaussian(img_cameraman, sigma=2)
img_edges = sobel(img_cameraman)
img_sharp = img_cameraman - ndimage.gaussian_filter(img_cameraman, sigma=3)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
titles = ['Original', 'Gaussian Blur (σ=2)', 'Sobel Edges', 'Sharpened']
images = [img_cameraman, img_blur, img_edges, img_sharp]

for ax, img, title in zip(axes, images, titles):
    ax.imshow(img, cmap='gray')
    ax.set_title(title, fontsize=10)
    ax.axis('off')

plt.suptitle('Image Filtering Comparison', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Task 4: Feature Extraction

```python
from skimage.feature import hog
from skimage import feature

# HOG features
hog_features, hog_image = hog(
    img_cameraman,
    orientations=8,
    pixels_per_cell=(16, 16),
    cells_per_block=(1, 1),
    visualize=True
)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_cameraman, cmap='gray')
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(hog_image, cmap='gray')
axes[1].set_title(f'HOG Features ({len(hog_features)} features)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
print(f'HOG feature vector length: {len(hog_features)}')
```
