## Task 1: Python & NumPy Basics

```python
import numpy as np
import matplotlib.pyplot as plt

# Basic array operations
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print('Array a:', a)
print('Array b:', b)
print('Sum:', a + b)
print('Dot product:', np.dot(a, b))
print('Mean of a:', np.mean(a))
print('Std of b:', np.std(b))
```

## Task 2: Matrix Operations

```python
# Matrix creation and operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('Matrix A:\n', A)
print('Matrix B:\n', B)
print('A + B:\n', A + B)
print('A @ B (matrix multiply):\n', A @ B)
print('Transpose of A:\n', A.T)
print('Determinant of A:', np.linalg.det(A))
print('Inverse of A:\n', np.linalg.inv(A))
```

## Task 3: Data Visualisation

```python
import pandas as pd

# Create sample dataset
np.random.seed(42)
data = pd.DataFrame({
    'x': np.linspace(0, 10, 100),
    'y_linear': np.linspace(0, 10, 100) + np.random.normal(0, 0.5, 100),
    'y_quad': np.linspace(0, 10, 100)**2 * 0.1 + np.random.normal(0, 0.5, 100)
})

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(data['x'], data['y_linear'], s=10, color='#0EA5E9', alpha=0.7)
axes[0].set_title('Linear Relationship')
axes[0].set_xlabel('x'); axes[0].set_ylabel('y')

axes[1].scatter(data['x'], data['y_quad'], s=10, color='#6366F1', alpha=0.7)
axes[1].set_title('Quadratic Relationship')
axes[1].set_xlabel('x'); axes[1].set_ylabel('y')

plt.tight_layout()
plt.show()
```

## Task 4: Data Preprocessing

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split

# Generate data with missing values
np.random.seed(42)
X = np.random.randn(100, 4)
X[np.random.choice(100, 10), np.random.choice(4, 10)] = np.nan

# Handle missing values
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
X_clean = imputer.fit_transform(X)

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_clean)

print(f'Original shape: {X.shape}')
print(f'Missing values: {np.isnan(X).sum()}')
print(f'After imputation - missing: {np.isnan(X_clean).sum()}')
print(f'Scaled mean: {X_scaled.mean():.6f} (should be ~0)')
print(f'Scaled std:  {X_scaled.std():.6f} (should be ~1)')
```
